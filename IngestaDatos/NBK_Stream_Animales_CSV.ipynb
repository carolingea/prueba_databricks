```python
(
    spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "csv")
        .option("header", "true")
        .option("cloudFiles.schemaLocation", "/Volumes/dbbosque/animal/metadatos/schema")
        .load("/Volumes/dbbosque/animal/archivos")
        .writeStream
        .option("checkpointLocation", "/Volumes/dbbosque/animal/metadatos/checkpoint")
        .option("mergeSchema", "true")
        .trigger(availableNow=True)
        .toTable("dbbosque.animal.animalesBronce")
)
```
