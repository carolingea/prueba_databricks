```python
%sql
-- VOLVER CLUSTER UNA TABLA (Liquid Clustering)
ALTER TABLE workspace.default.products_2000000 CLUSTER BY (Index);

-- CREAR UNA TABLA CLUSTER A PARTIR DE OTRA
CREATE TABLE workspace.default.products_200000_cluster 
USING DELTA
CLUSTER BY AUTO
AS 
SELECT 
    Index as Id, Name, Description,
    Brand, Category, Price, Currency,
    Stock, EAN, Color, 
    Size, Availability
FROM workspace.default.products_2000000

```

```python
%sql
-- OPTIMIZAR
OPTIMIZE workspace.default.products_200000_cluster;
```

```python
%sql

-- Verificar que quedo configurado el cluster
DESCRIBE DETAIL workspace.default.products_200000_cluster;

-- VER HISTORIAL
DESCRIBE HISTORY workspace.default.products_200000_cluster;


```
